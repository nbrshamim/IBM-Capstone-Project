### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
#file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
file_path ="survey_data_duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
#duplicates_data = df.duplicated().sum()
# Identify the duplicate rows
duplicates_data = df.duplicated().sum()
print("Total Duplicate data:")
print(duplicates_data)

duplicates = df.duplicated()# Display the duplicate rows
print("Duplicate Rows:")
print(df[duplicates])

```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
df_cleaned = df.drop_duplicates()

dropped_count = df.shape[0] - df_cleaned.shape[0]
print(f"Successfully removed {dropped_count} duplicate rows.")

# Display the DataFrame after removing duplicates
print("DataFrame after removing duplicates:")
print(df_cleaned.head())

df_cleaned = df_cleaned.duplicated().sum()
print("After remove Duplicate, Total duplicatdata:",df_cleaned)
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
#find the missing values for all columns
missing_values = df.isnull().sum()
# Display the missing values for each column   
print("Missing values in each column:")
print(missing_values)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
missing_remote_work = df['RemoteWork'].isnull().sum()

print(f"Number of missing values in 'RemoteWork' column: {missing_remote_work}")
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
remotework_count = df['RemoteWork'].value_counts()
remotework_count
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
max_frequent_remotework = remotework_count.idxmax()
max_frequent_remotework

## It also work
#most_frequent_value = df['RemoteWork'].mode()[0]
#most_frequent_value
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
#df['RemoteWork'] = df['RemoteWork'].fillna(max_frequent_remotework, inplace=True)
df.fillna({'RemoteWork': max_frequent_remotework}, inplace=True)

# Display the DataFrame after imputation
print("DataFrame after imputing 'RemoteWork':")
print(df.head())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
#Check for any compensation-related columns and describe their distribution.
# compensation_columns = [col for col in df.columns if 'Comp' in col]
# if compensation_columns:
#     for col in compensation_columns:
#         print(f"Distribution of {col}:")
#         print(df[col].describe())
# else:
#     print("No compensation-related columns found.")
compensation_columns = [col for col in df.columns if 'Comp' in col or 'Salary' in col]
print("Compensation-related columns:")
print(compensation_columns)

if compensation_columns:
    for col in compensation_columns:
        print(f"\nSummary statistics for {col}:")
        print(df[col].describe())
else:
    print("No compensation-related columns found.")
```
